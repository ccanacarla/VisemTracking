```python
PROJECT_ROOT = Path.cwd().parent

DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
DATA_PROCESSED.mkdir(parents=True, exist_ok=True)

DATA_SYMBOLIC = PROJECT_ROOT / "data" / "symbolic"
DATA_SYMBOLIC.mkdir(parents=True, exist_ok=True)

print("Processed data folder:", DATA_PROCESSED)
```

```python
df = pd.read_csv(DATA_PROCESSED / "02_processed.csv")
df
```

```python
# ======================================================
# Discretização da velocidade em 5 níveis (Likert)
# SEM classe "Parado"
# ======================================================

# Remove NaN apenas para cálculo dos limiares
valid_speed = df["speed"].dropna()

# Quintis da velocidade
if not valid_speed.empty:
    q20, q40, q60, q80 = np.quantile(
        valid_speed,
        [0.20, 0.40, 0.60, 0.80]
    )
else:
    q20 = q40 = q60 = q80 = 0.0


# -----------------------------
# Rotulagem da velocidade
# -----------------------------
def label_speed_likert(v):
    if np.isnan(v):
        return np.nan
    if v <= q20:
        return "Muito_Lento"
    if v <= q40:
        return "Lento"
    if v <= q60:
        return "Medio"
    if v <= q80:
        return "Rapido"
    return "Muito_Rapido"


# Aplica rotulagem (vetorizado)
df["speed_label"] = df["speed"].map(label_speed_likert)
#df["speed_label"] = ""
```

Direção

```python
def label_dir(a):
    if np.isnan(a):
        return np.nan

    # Normaliza ângulo para [-180, 180]
    a = ((a + 180) % 360) - 180

    if -45 < a <= 45:
        return "Leste"
    if 45 < a <= 135:
        return "Norte"
    if a > 135 or a <= -135:
        return "Oeste"
    return "Sul"

df["dir4_label"] = df["angle_deg"].map(label_dir)
#df["dir4_label"] = ""
```

Giro

```python
# Diferenças em x e y por trajetória
df['dx'] = df.groupby('trajectory_id')['x'].diff()
df['dy'] = df.groupby('trajectory_id')['y'].diff()

# Distância Euclidiana
df['delta_dist'] = np.sqrt(df['dx']**2 + df['dy']**2)

# Primeiro ponto da trajetória não tem deslocamento
df['delta_dist'] = df['delta_dist'].fillna(0)

# Ignora zeros (frames parados) para não distorcer
dist_vals = df.loc[df['delta_dist'] > 0, 'delta_dist']

q = dist_vals.quantile([0.2, 0.4, 0.6, 0.8]).values
q1, q2, q3, q4 = q

def discretize_distance_5_quantile(d):
    if d == 0:
        return "D0"
    elif d <= q1:
        return "D1"
    elif d <= q2:
        return "D2"
    elif d <= q3:
        return "D3"
    else:
        return "D4"

df['dist_label'] = df['delta_dist'].apply(discretize_distance_5_quantile)
```

Linearidade

```python
window = 5

df['step_dist'] = np.sqrt(df.groupby('trajectory_id')['x'].diff()**2 + 
                              df.groupby('trajectory_id')['y'].diff()**2).fillna(0)
    
# Soma da distância percorrida na janela (VCL local)
rolling_path = df.groupby('trajectory_id')['step_dist'].rolling(window, center=True).sum().reset_index(0, drop=True)

# Distância vetorial (VSL local) - simplificação rápida aproximada
# Pega o deslocamento total em X e Y na janela
roll_dx = df.groupby('trajectory_id')['x'].diff(window).shift(-int(window/2))
roll_dy = df.groupby('trajectory_id')['y'].diff(window).shift(-int(window/2))
rolling_dist = np.sqrt(roll_dx**2 + roll_dy**2)

# Cálculo da Linearidade (0 a 1)
df['local_lin'] = (rolling_dist / rolling_path).fillna(1.0) # 1.0 se parado

TH_LOW = 0.5
TH_HIGH = 0.85

def label_linearity(lin):
    if np.isnan(lin): return "NA"
    if lin >= TH_HIGH:
        return "Linear"      # Movimento progressivo reto
    if lin >= TH_LOW:
        return "Curvo"       # Movimento em arco aberto
    return "Caotico"         # Hiperativo, vibratório ou círculo fechado

df['lin_label'] = df['local_lin'].map(label_linearity)
    
```

Combinação simbólica

```python
def combine_labels_v2(speed, direction, lin):
    if pd.isna(speed) or speed == "Parado":
        return "Parado"
    # Exemplo de símbolo: "Rapido_Norte_Linear" ou "Muito_Rapido_Sul_Caotico"
    return f"{speed}_{direction}_{lin}"

df["symbol"] = [
    combine_labels_v2(s, d, l)
    for s, d, l in zip(df["speed_label"], df["dir4_label"], df["dist_label"])
]

df.to_csv(DATA_SYMBOLIC / "01_symbolic.csv", index=False)
df
```

Sequência simbólica por trajetória

```python
symbolic = (
    df
    .sort_values("timestamp")
    .groupby("trajectory_id")["symbol"]
    .agg(list)
    .reset_index(name="symbolic_movement")
)

# Salva resultado
symbolic.to_csv(DATA_SYMBOLIC / "02_symbolic_per_trajectory.csv", index=False)
```
